Missing values

```python
def missing_report(df, name):

    print(f"\n{'=' * 70}")
    print(name.upper())
    print("=" * 70)

    result = []

    for col in df.columns:
        empty = (df[col].astype(str).str.strip() == "").sum()

        result.append({
            "column": col,
            "empty_count": empty,
            "empty_percentage": round(
                empty / len(df) * 100, 4
            )
        })

    return pd.DataFrame(result)


for name, df in train_data.items():
    if name != "ground_truth":
        display(missing_report(df, name))
```

```python
for name, df in test_data.items():
    display(missing_report(df, f"test_{name}"))
```

Language/Script analysis

```python
import re


def detect_script(text):

    text = str(text)

    if re.search(r"[\u0900-\u097F]", text):
        return "Devanagari"

    if re.search(r"[\u0B80-\u0BFF]", text):
        return "Tamil"

    if re.search(r"[\u0C00-\u0C7F]", text):
        return "Telugu"

    if re.search(r"[\u0C80-\u0CFF]", text):
        return "Kannada"

    if re.search(r"[\u0D00-\u0D7F]", text):
        return "Malayalam"

    if re.search(r"[\u0600-\u06FF]", text):
        return "Arabic"

    if re.search(r"[A-Za-z]", text):
        return "Latin"

    return "Other"
```

```python
sample = train_data["source1"]["business_name"].head(10000)

script_counts = sample.map(detect_script).value_counts()

print(script_counts)
```

```python
source1 = train_data["source1"]

sample_source1 = source1.head(10000).copy()

sample_source1["script"] = (
    sample_source1["business_name"]
    .map(detect_script)
)

for script in sample_source1["script"].unique():

    print(f"\n{'=' * 60}")
    print(f"SCRIPT: {script}")
    print("=" * 60)

    display(
        sample_source1.loc[
            sample_source1["script"] == script,
            [
                "entity_id",
                "business_name",
                "business_address",
                "country",
                "script"
            ]
        ].head(10)
    )
```
